Separando intevalos de tempo

```python
# Separa usando Regex que aceita 'às', 'ás' ou 'as' com espaços ao redor
df_analise[['INICIO_GRAVACAO', 'FIM_GRAVACAO']] = df_analise['PERIODO DE GRAVAÇÃO'].str.split(r'\s+[aàá]s\s+', expand=True, regex=True)

# Remove espaços extras nas pontas
df_analise['INICIO_GRAVACAO'] = df_analise['INICIO_GRAVACAO'].str.strip()
df_analise['FIM_GRAVACAO'] = df_analise['FIM_GRAVACAO'].str.strip()

# Exibe o resultado corrigido
df_analise[['PERIODO DE GRAVAÇÃO', 'INICIO_GRAVACAO', 'FIM_GRAVACAO']].head()
```

Expressões regulares na coluna ação

```python
# Regex flexível para acentos (saída/saida), múltiplos espaços e textos longos no final
padrao_regex = r"Acompanhado pelo vigilante\s+(.+?)\s+até a sa[íi]da\s+(.+)"

# Aplica a extração corrigida
df_analise[['VIGILANTE', 'SAIDA']] = df_analise['AÇÃO TOMADA'].str.extract(padrao_regex, expand=True)

# Limpa os espaços em branco que sobraram nas pontas
df_analise['VIGILANTE'] = df_analise['VIGILANTE'].str.strip()
df_analise['SAIDA'] = df_analise['SAIDA'].str.strip()

# Exibe o resultado corrigido
df_analise[['AÇÃO TOMADA', 'VIGILANTE', 'SAIDA']].head()
```

```python
# 1. Transforma o que for texto para String e remove pontos finais/espaços nas pontas
df_analise['VIGILANTE'] = df_analise['VIGILANTE'].astype(str).str.replace(r'\s*\.\s*$', '', regex=True).str.strip()

# 2. Padroniza para Capitalize (Primeira letra de cada palavra em maiúscula)
df_analise['VIGILANTE'] = df_analise['VIGILANTE'].str.title()

# 3. Trata os valores nulos que viraram a string 'Nan' de volta para o nulo real do Pandas
import numpy as np
df_analise['VIGILANTE'] = df_analise['VIGILANTE'].replace({'Nan': np.nan, '': np.nan})
```

```python
df_analise
```

```python
# 1. Converter DATA para o tipo datetime do Pandas
df_analise['DATA'] = pd.to_datetime(df_analise['DATA'], errors='coerce')

# 2. Converter horários nativos para string e depois para Timedelta de forma segura
for col in ['HORÁRIO', 'INICIO_GRAVACAO', 'FIM_GRAVACAO']:
    df_analise[col] = pd.to_timedelta(df_analise[col].astype(str), errors='coerce')

# 3. Converter colunas com textos repetitivos para o tipo 'category' (Corrigido para N° CÂMERA)
colunas_categoricas = ['TURNO', 'LOCAL DE IDENFICAÇÃO', 'TIPO DE OCORRÊNCIA', 'N° CÂMERA', 'VIGILANTE', 'SAIDA']
for col in colunas_categoricas:
    df_analise[col] = df_analise[col].astype('category')

# 4. Verificar se a nova estrutura de tipos foi aplicada com sucesso
df_analise.info()
```

```python
# 1. Calcula a duração real (subtraindo o início do fim)
df_analise['DURACAO_GRAVACAO'] = df_analise['FIM_GRAVACAO'] - df_analise['INICIO_GRAVACAO']

# 2. Cria uma versão visual limpa dos horários (sem o "0 days") apenas para exibição
df_visualizacao = df_analise[['INICIO_GRAVACAO', 'FIM_GRAVACAO', 'DURACAO_GRAVACAO']].copy()

for col in df_visualizacao.columns:
    # Extrai horas, minutos e segundos e formata como HH:MM:SS
    horas = df_visualizacao[col].dt.components['hours'].astype(str).str.zfill(2)
    minutos = df_visualizacao[col].dt.components['minutes'].astype(str).str.zfill(2)
    segundos = df_visualizacao[col].dt.components['seconds'].astype(str).str.zfill(2)
    df_visualizacao[col] = horas + ':' + minutos + ':' + segundos

# 3. Exibe o resultado limpo na tela
df_visualizacao.head()
```
